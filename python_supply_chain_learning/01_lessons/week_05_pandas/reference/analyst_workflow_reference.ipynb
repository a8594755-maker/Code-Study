## 1. 載入工具

```python
import pandas as pd
```

## 2. 讀 CSV

人類思路：我先把資料讀進 Python，變成一張 DataFrame。

```python
from pathlib import Path

data_file = Path("/Users/xuweijin/Code study/python_supply_chain_learning/03_data/superstore.csv")

df = pd.read_csv(data_file)
```

## 5. 檢查資料型態

人類思路：我想知道哪些欄位是數字、哪些是文字、哪些可能需要轉成日期。

```python
df.info()
```

## 6. 第一個分析問題：哪個產品類別最賺錢？

人類思路：我要按 Category 分組，然後加總 Sales 和 Profit。

```python
category_summary = df.groupby("Category")[["Sales", "Profit"]].sum()

category_summary
```

## 7. 把最賺錢的排在最上面

人類思路：表格有了，但我要按照 Profit 從高到低排序，這樣比較容易看結論。

```python
category_summary = category_summary.sort_values("Profit", ascending=False)

category_summary
```

## 8. 第二個分析問題：哪個地區最賺錢？

人類思路：這次我要把分組欄位從 Category 換成 Region。

```python
region_summary = (
    df.groupby("Region")[["Sales", "Profit"]]
    .sum()
    .sort_values("Profit", ascending=False)
)

region_summary
```

## 9. 第三個分析問題：哪些訂單在虧錢？

人類思路：我要篩選 Profit 小於 0 的資料，再看虧損最大的訂單。

```python
loss_orders = df[df["Profit"] < 0]

loss_orders[["Order ID", "Category", "Sub-Category", "Product Name", "Sales", "Discount", "Profit"]].sort_values("Profit").head(10)
```

## 10. 用人話整理結論

人類思路：分析不是只印表格，最後要把結果變成可以被理解的句子。

```python
best_category = category_summary.index[0]
best_category_profit = category_summary.iloc[0]["Profit"]

best_region = region_summary.index[0]
best_region_profit = region_summary.iloc[0]["Profit"]

print(f"Best category by profit: {best_category} (${best_category_profit:,.2f})")
print(f"Best region by profit: {best_region} (${best_region_profit:,.2f})")
```
